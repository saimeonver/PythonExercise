```python
import pandas as pd
import seaborn as sns
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 500)
df = sns.load_dataset("titanic")
df.head()
```

```python
df["age2"] = df["age"] * 2
```

```python
df["age3"] = df["age"] * 3
```

```python
#her biri için tek tek işlem yapacağıma döngü yazarım
for col in df.columns:
  if "age" in col:
    print(col)
```

```python
for col in df.columns:
  if "age" in col:
    print((df[col]/10).head())
```

```python
for col in df.columns:
  if "age" in col:
    df[col]=df[col]/10
```

```python
df.head()
```

```python
#apply ile nasıl yaparım
df[["age", "age2" , "age3"]].apply(lambda x : x/10).head()
```

```python
df.loc[:, df.columns.str.contains("age")].apply(lambda x : x/10).head()
```

```python
#değerleri standartlaştırdık
df.loc[:, df.columns.str.contains("age")].apply(lambda x: (x - x.mean()) / x.std())
```

```python
#değerleri standartlaştırmanın diğer yolu

def standart_scale(col_name):
  return (col_name - col_name.mean()) / col_name.std()

df.loc[: , df.columns.str.contains("age")].apply(standart_scale).head()
```

```python
#kalıcı şekilde değişikliği gerçekleştirdim
df.loc[:, ["age", "age2", "age3"]]= df.loc[:, df.columns.str.contains("age")].apply(standart_scale)
```

```python
#yukarıdaki işlemin diğer yolu
df.loc[:, df.columns.str.contains("age")]= df.loc[:, df.columns.str.contains("age")].apply(standart_scale)
df.head()
```

```python
#birleştirme (Join) işlemleri

import numpy as np
import pandas as pd
m = np.random.randint(1, 30, size = (5,3))
df1 = pd.DataFrame(m, columns = ["var1", "var2", "var3"])
df2 = df1 + 99
```

```python
pd.concat([df1,df2]) # iki dataframi birleştirdim
```

```python
pd.concat([df1,df2], ignore_index = True)
```

```python
#Merge ile birleştirme işlemleri

df1 = pd.DataFrame({'employees': ['john', 'dennis','mark','maria'],
                    'group': ['accounting','engineering','enginnering','hr']})
df2 = pd.DataFrame({'employees': ['mark','john','dennis','maria'],'start_date':[2010,2009,2014,2019]})
```

```python
df1
```

```python
df2
```

```python
pd.merge(df1,df2)
pd.merge(df1, df2, on='employees')
```

```python
#amaç : her çalışanın müdürünün bilgisine erişmek istiyoruz
df3 = pd.merge(df1,df2)
```
