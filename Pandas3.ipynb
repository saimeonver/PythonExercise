```python
#Koşullu Seçim (Conditional Selection)

import pandas as pd
import seaborn as sns
pd.set_option('display.max_columns',None)
df = sns.load_dataset("titanic")
df.head()
```

```python
df[df["age"] > 50].head()
```

```python
df[df["age"] >50].count()
```

```python
df[df["age"] > 50]["age"].count()
```

```python
df.loc[df["age"] > 50, "class"].head()
```

```python
df.loc[df["age"] >50,"class"].head() # bir koşul bir sütun seçtim
```

```python
df.loc[df["age"] >50, ["age","class"]].head() #bir koşul iki sütun
```

```python
#ikinci  koşulu eklerken araya & konur iki koşulun dışına da ayrı ayrı parantez konur
df.loc[(df["age"]>50) & (df["sex"] == "male"), ["age","class"]].head()
```

```python
df.loc[(df["age"]>50 )
& (df["sex"]== "male")
& (df["embark_town"] == "Cherbourg"),
  ["age","class","embark_town"]].head()
```

```python
df["embark_town"].value_counts()
```

```python
df_new = df.loc[( df["age"] > 50 )
& (df["sex"] == "male")
& ((df["embark_town"] == "Cherbourg") | (df["embark_town"]== "Southampton")),
["age","class","embark_town"]].head()

# | ya da demek , & ve demek
```

```python
df_new["embark_town"].value_counts()
```

```python
import pandas as pd
import seaborn as sns
pd.set_option('display.max_columns',None)
df = sns.load_dataset("titanic")
df.head()
```

```python
df["age"].mean()
```

```python
df.groupby("sex")["age"].mean()
```

```python
df.groupby("sex").agg({"age":"mean"})
```

```python
df.groupby("sex").agg({"age" : ["mean","sum"]})
```

```python
df.groupby("sex").agg({"age":["sum","mean"],"embark_town": "count"})
#burada embark town detayları cıkmadı cinsiyete göre sayı hesapladı.
#pivot kullanmak lazım
```

```python
df.groupby("sex").agg({"age":["sum","mean"],"survived": "mean"})
#kadınların bir çoğu hayatta kalmış erkeklere göre
```

```python
df.groupby(["sex","embark_town"]).agg({"age":"mean", "survived":"mean"})
```

```python
df.groupby(["sex","embark_town","class"]).agg({"age":"mean", "survived":"mean"})
```

```python
df.groupby(["sex","embark_town","class"]).agg({"age":"mean", "survived":"mean","sex":"count"})
```

```python
from pandas._config.config import set_option
#pivot table
import pandas as pd
import seaborn as sns
pd.set_option('display.max_columns',None)
df = sns.load_dataset("titanic")
df.head()
```

```python
df.pivot_table("survived","sex","embarked")
#ilki kesişimi verir, ikinci satırları , üçüncü sütunları
```

```python
df.pivot_table("survived","sex","embarked",aggfunc="std")
```

```python
df.pivot_table("survived","sex",["embarked","class"])
```

```python
#cut ve qcut sayısal değişkenleri katogorik değkene çevirir
#cut hangi kategoiye böleceğin belliyse
#qcut ise çeyreklik bölümlere ayırır
#yaş değişkennini yorumladığım içn cut ı kullanırım 60 yaş üstü yaşlı dedim mesela
df["new_age"]=pd.cut(df["age"],[0, 10, 18, 25, 40, 90])
```

```python
df.pivot_table("survived", "sex", "new_age")
```

```python
df.pivot_table("survived", "sex", ["new_age","class"])
```
