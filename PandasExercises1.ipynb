```python
#Pandas Alıştırmalar
#1 seaborn kütüphanesinden titanic veri setini tanımla
import pandas as pd
import seaborn as sns
df = sns.load_dataset("titanic")
df.head()
```

```python
df["sex"].value_counts()
```

```python
#herbir sütuna ait unique değerleri bulun
df.nunique()
```

```python
from numpy import unique_values
#görev 4: pclass değişkeninin unique değerlerinin sayısını bul
df['pclass'].nunique()
```

```python
#görev 5: pclass ve parch değişkenlerinin unique değerlerinin
#sayısını bul
df[['pclass','parch']].nunique()
```

```python
#görev6 : embarked değişkenin tipini bulun. tipini category olarak değiştir
#ve tekrar kontrol et
df["embarked"].dtype
```

```python
df["embarked"] = df["embarked"].astype("category")
```

```python
df["embarked"].dtype
```

```python
#grev 7 : embarked değeri C olanların tüm bilgilerini gösterin

df.loc[df["embarked"] == "C"]
```

```python
#görev 8: embarked değeri S olmaynaları göster

df.loc[df["embarked"] != 'S']
```

```python
df[df["embarked"] != "S"]["embarked"].unique()
```

```python
#görev 9 : Yaşı 30 dan küçük ve kadın olan yolcuların tüm bilgilerini göster
df.loc[(df["sex"] == "female") & (df["age"] < 30)]
```

```python
#fare'i 500 den büyük veya yaşı 70 den büyük yolcularaın bilgilerini göster

df.loc[(df["fare"] > 500) | (df["age"] > 70)]
```

```python
#görev 11 : her bir değişkende boş değerlerin toplamını bul

df.isnull().sum()
```

```python
#görev 12 : who değişenini dataframeden çıkar

df.drop("who",axis = 1, inplace= True)
```

```python
df.head()
```

```python
#görev 13: deck değişkenindeki boş değerleri deck değişkeninin en çok tekrar
#eden (mod) ile doldur

```

```python
mode_value = df["deck"].mode()[0]
df["deck"] = df["deck"].fillna(mode_value)
df["deck"].isnull().sum()
```
