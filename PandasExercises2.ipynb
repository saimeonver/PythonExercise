```python
#görev 14: age değişkenindeki boş değerleri age değişkeninin
#medyanı ile doldurun

import seaborn as sns
import pandas as pd
df = sns.load_dataset("titanic")
df.head()
```

```python
age_medyan = df["age"].median()
df["age"] = df["age"].fillna(age_medyan)
```

```python
df[df["age"].isnull()]
```

```python
#görev 15: survived deişkneinin pclass ve cinsiyet değişkenleri kırılımnda
#sum, count, mean değerlerini bulun
df.pivot_table("survived", ["pclass", "sex"], aggfunc=["sum","count","mean"])
```

```python
#görev 16: 30 yaş aştında olanlar 1, 30 a eşit olanlar 0 verecek fonks yaz
#yazdığnda fonsiyonda titanic veri setinde age_flag adndan bir değişken
#oluştur(apply ve lambda kullan)

def age_flag_func(x):
    if x < 30:
        return 1
    else:
        return 0



df["age_flag"] = df["age"].apply(lambda x: age_flag_func(x) )
```

```python
#diğer yol
df["age_flag"] = df["age"].apply(lambda x:1 if x<30 else 0 )
```

```python
#görev 17: seaborn kütüphanesinden tips veerisini tanımlayın
import seaborn as sns
import pandas as pd

df = sns.load_dataset("tips")
df.head()
```

```python
df.shape
```

```python
#görev 18: time değişkenini kategorilerine (dinner ,lunch) a göre total_bill
#değerlerinin toplamoıını, min,max,ortalamasını bulunuz

df.groupby("time").agg({"total_bill":["sum","min","max","mean"]})
```

```python
#görev 19: günlere ve time a göre total_bill değerinin toplamını
#min, max ve ortalmasını bulun

df.groupby(["day","time"]).agg({"total_bill": ["sum","min","max","mean"]})
```

```python
#görev 20: Lunch zamanına göre ve kadın müşterilere ait total_bill ve tip değer
#lerinin day 2 e göre toplamnın min, max ve ortalamsını bulunuz

df_new = df.loc[(df["sex"]=="Female")&(df["time"]=="Lunch")]
```

```python
df_new.groupby("day").agg({"total_bill": ["sum","min","max","mean"],
                           "tip": ["sum","min","max","mean"]})
```

```python
#görev21: sizei 3 ten küçük total_bill 10'dan büyük olan sipraişlerin
#ortalaması nedir(loc kullan)

df.loc[(df["size"]<3) & (df["total_bill"]>10), "total_bill"].mean()
```

```python
#görev22  :total_bill_tip_sum adından yeni bir dğeişken oluştur. her bir
#müşterinin ödediği totalbill ve tp in toplamını ver

df["total_bill_tip_sum"] = df["total_bill"] + df["tip"]
df.head()
```

```python
#görev23 : total_bill_tip_sum değişkeine göre büyükten küçüğe sırala
#ilk 30 kişiyi yen df e ata

df_new_30 = df["total_bill_tip_sum"].sort_values(ascending=False).head(30)
```
